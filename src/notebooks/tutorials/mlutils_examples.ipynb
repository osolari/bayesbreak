```python
from src.estimator.mlutils import est_profile_with_mbpcr


results = est_profile_with_mbpcr(
    path='',
    sample_name='rec10k',
    snp_name=rec10k['snp_name'],
    chr=rec10k['chr'],
    position=rec10k['position'],
    logratio=rec10k['logratio'],
    chr_to_be_analyzed=list(range(1, 23)) + ['X'],
    max_probe_number=2000
)
```

```python
import matplotlib.pyplot as plt

# Assuming import_cn_data and compute_mbpcr are already implemented

# Import the CN data
jekoChr11Array250Knsp = import_cn_data(os.path.join(DATA_RESOURCES_DIR, "jekoChr11Array250Knsp.tsv"), n_row_skip=1)

# Select a part of chromosome 11
y = jekoChr11Array250Knsp['logratio'][10600:11600]
p = jekoChr11Array250Knsp['position'][10600:11600]
```

```python
from bprseg.estimator.mlutils import computeMBPCR


results = computeMBPCR(
    y,
    nu=-3.012772e-10,
    rho_square=0.0479,
    sigma_square=0.0699,
    regr=2
)
```

```python
# Plot the data
plt.plot(p, y, 'o', color='grey', label='Logratio')
plt.plot(p, results['estPC'], color='red', label='Estimated Profile')
plt.plot(p, results['regrCurve'], color='green', label='Regression Curve')
plt.xlabel('Position')
plt.ylabel('Logratio')
plt.title('Estimated Profile of Part of Chromosome 11')
plt.legend()
plt.show()
```
